```python
import time
from anthropic import Anthropic
client = Anthropic()
models = ["claude-haiku-4-5", "claude-sonnet-4-5", "claude-opus-4-5"]
prompt = "Summarize the French Revolution in 3 sentences."
for m in models:
    t0 = time.time()
    r = client.messages.create(model=m, max_tokens=150, messages=[{"role": "user", "content": prompt}])
    dt = time.time() - t0
    print(m, dt, r.usage.input_tokens, r.usage.output_tokens)
```
